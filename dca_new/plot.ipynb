```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Given data dictionary
data = {
    'index': ['FPT_2020', 'FPT_2021', 'FPT_2022', 'FPT_2023', 'FPT_2024',
              'E1VFVN30_2020', 'E1VFVN30_2021', 'E1VFVN30_2022', 'E1VFVN30_2023', 'E1VFVN30_2024'],
    'stock_symbol': ['FPT', 'FPT', 'FPT', 'FPT', 'FPT', 
                     'E1VFVN30', 'E1VFVN30', 'E1VFVN30', 'E1VFVN30', 'E1VFVN30'],
    'Total investment': [1191.6, 2135.7, 3454.3, 4582.3, 5984.2, 
                         1177.2, 2238.8, 3580.7, 4655.4, 5986.0],
    'Total value': [1491.0, 3015.0, 7241.0, 8463.0, 14137.2, 
                       1180.8, 2880.0, 5682.6, 4679.1, 6694.6],
}

# Create DataFrame
df = pd.DataFrame(data)

# Adjust currency units (multiply by 10,000)
df['Total investment'] /= 100
df['Total value'] /= 100

# Melt the DataFrame to long format for seaborn
df_melted = df.melt(id_vars=['stock_symbol', 'index'], 
                    value_vars=['Total investment', 'Total value'],
                    var_name='Metrics', value_name='amount')

# Extract year for plotting
df_melted['year'] = df_melted['index'].str.split('_').str[1]

# Create the bar plot
plt.figure(figsize=(12, 6))
ax= sns.barplot(x='year', y='amount', hue='Metrics', data=df_melted[df_melted['stock_symbol'].isin(['FPT'])],
            palette='colorblind', errorbar=None)
ax.bar_label(ax.containers[0], fmt='%.2f', fontsize=14)
ax.bar_label(ax.containers[1], fmt='%.2f', fontsize=14)
# Setting titles and labels
plt.xlabel('') 
plt.ylabel('Million VND', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='', fontsize=12)
#plt.title('Total Investment and Total Value for FPT Over the Years', fontsize=16)
# Show the plot
plt.grid(False)
plt.tight_layout()
plt.show()
```
